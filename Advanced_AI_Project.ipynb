# Load Libraries and Data Source

```python
# ---------------------------------------------
# 1. Import required libraries
# ---------------------------------------------
import os
import librosa
import librosa.display
import numpy as np
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

# For audio playback in notebooks (optional)
from IPython.display import Audio

# ---------------------------------------------
# 2. Set the path to your RAVDESS dataset
#    Update this to wherever you unzipped the Kaggle dataset
# ---------------------------------------------
DATA_DIR = Path("/DataSet")

# ---------------------------------------------
# 3. Collect all .wav files (there should be 1,440)
# ---------------------------------------------
audio_files = sorted(list(DATA_DIR.rglob("*.wav")))
print(f"Found {len(audio_files)} audio files.")

# ---------------------------------------------
# 4. Load a single example file to verify things work
# ---------------------------------------------
example_path = audio_files[0]
print("Example file:", example_path)

# Load audio (default sampling rate = 22050 Hz)
y, sr = librosa.load(example_path, sr=None)  # sr=None preserves original sampling rate

print(f"Audio duration: {len(y)/sr:.2f} seconds")
print(f"Sampling rate: {sr} Hz")

# ---------------------------------------------
# 5. Display waveform and play audio (optional)
# ---------------------------------------------
plt.figure(figsize=(10, 3))
librosa.display.waveshow(y, sr=sr)
plt.title("Waveform")
plt.show()

Audio(example_path)
```
